```python
import os
from PIL import Image
import pandas as pd
import matplotlib.pyplot as plt
from collections import defaultdict

# Define the directories to analyze
directories = [
    "/home/stat/lee/DL2/data/imagenette_160/train/n01440764",
    "/home/stat/lee/DL2/data/mini_coco_det/train",
    "/home/stat/lee/DL2/data/mini_voc_seg/train"
]

# Dictionary to store image size information for each directory
image_sizes = {}

# Extensions to check
valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff')

# Process each directory
for directory in directories:
    dir_name = os.path.basename(directory)
    sizes = []
    sizes_dict = defaultdict(int)
    
    if not os.path.exists(directory):
        print(f"Directory does not exist: {directory}")
        continue
        
    print(f"Analyzing {directory}...")
    
    # Get all image files in the directory (including subdirectories)
    image_files = []
    for root, _, files in os.walk(directory):
        for file in files:
            if file.lower().endswith(valid_extensions):
                image_files.append(os.path.join(root, file))
    
    # Process each image file
    for img_path in image_files[:100]:  # Limit to first 100 for speed
        try:
            with Image.open(img_path) as img:
                width, height = img.size
                sizes.append((width, height))
                sizes_dict[f"{width}x{height}"] += 1
        except Exception as e:
            print(f"Error processing {img_path}: {e}")
    
    # Store results
    image_sizes[dir_name] = {
        "count": len(image_files),
        "unique_sizes": len(sizes_dict),
        "sizes": sizes_dict,
        "samples": sizes[:10]  # Store first 10 sizes for reference
    }
    
    print(f"Found {len(image_files)} images with {len(sizes_dict)} unique sizes")

# Display results
for dir_name, data in image_sizes.items():
    print(f"\n=== {dir_name} ===")
    print(f"Total images: {data['count']}")
    print(f"Unique sizes: {data['unique_sizes']}")
    print("Most common sizes:")
    for size, count in sorted(data['sizes'].items(), key=lambda x: x[1], reverse=True)[:5]:
        print(f"  {size}: {count} images")

# Plot size distributions for comparison
# plt.figure(figsize=(15, 5))
# for i, (dir_name, data) in enumerate(image_sizes.items(), 1):
#     plt.subplot(1, len(image_sizes), i)
    
#     # Extract width and height from all samples
#     widths = [s[0] for s in data.get('samples', [])]
#     heights = [s[1] for s in data.get('samples', [])]
    
#     # Create scatter plot
#     plt.scatter(widths, heights, alpha=0.5)
#     plt.title(f"{dir_name}\n{data['count']} images, {data['unique_sizes']} sizes")
#     plt.xlabel('Width (pixels)')
#     plt.ylabel('Height (pixels)')
    
# plt.tight_layout()
# plt.show()

# Create a summary dataframe
summary = pd.DataFrame([
    {"Directory": d, 
     "Total Images": image_sizes.get(os.path.basename(d), {}).get("count", 0),
     "Unique Sizes": image_sizes.get(os.path.basename(d), {}).get("unique_sizes", 0),
     "Common Sizes": ", ".join([f"{s}" for s, c in sorted(
         image_sizes.get(os.path.basename(d), {}).get("sizes", {}).items(), 
         key=lambda x: x[1], reverse=True)[:3]])
    } 
    for d in directories
])

summary
```
